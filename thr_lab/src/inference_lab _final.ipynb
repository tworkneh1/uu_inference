## Set Up Code:

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

rng = np.random.default_rng(100)

def purchase_probability(price):
    price = np.asarray(price, dtype=float)
    return (
        0.55 * np.exp(-np.logaddexp(0.0, (price - 13.9) / 0.8))
        + 0.30 * np.exp(-np.logaddexp(0.0, (price - 30.9) / 1.2))
        + 0.15 * np.exp(-np.logaddexp(0.0, (price - 58.4) / 1.8))
    )

def sales(price=14.0, rng=None, mean_customers=30):
    price = float(price)
    if not np.isfinite(price):
        raise ValueError("price must be finite")
    if price < 0:
        raise ValueError("price must be nonnegative")
    mean_customers = float(mean_customers)
    if not np.isfinite(mean_customers) or mean_customers < 0:
        raise ValueError("mean_customers must be finite and nonnegative")
    rng = np.random.default_rng(rng)
    customer_count = rng.poisson(mean_customers)
    probability = purchase_probability(price)
    return rng.binomial(1, probability, size=customer_count).astype(np.int8)
```

# First set of experiments:

- Perform 14 experiments: the first uses a price of 5 and aims to experiment on 300 customers. Each subsequent experiment increases the price by 5 on ~300 customers until we perform our upper experiment using a price of 70
- We then calculate the total purchases, total revenue and revenue per customer for each experiment
- We can plot the revenue per customer vs price (aka each experiment) to get a visual sense of which price ranges performed the best (revenue wise)
- This first set of experiments is broad and not granular: we are looking for general trends and peaks to get a general sense of the optimal price range
- We will hone in a smaller range of prices to perform our second set of experiments using the data from this first set

```python
# Prices to experiment with
prices = np.arange(5, 70, 5)

# Expected customers at each price
experiment_size = 300

results = []
total_customers = 0
total_revenue = 0

for price in prices:

    purchases = sales(
        price=price,
        mean_customers=experiment_size,
        rng=rng
    )

    customers = len(purchases)
    revenue = purchases.sum() * price
    revenue_per_customer = revenue / customers


    total_customers += customers
    total_revenue += revenue

    results.append({
        "price": price,
        "customers": customers,
        "revenue": revenue,
        "revenue_per_customer": revenue_per_customer,
        'purchases': purchases.sum()
    })

results_df = pd.DataFrame(results)

best_row = results_df.loc[
    results_df["revenue_per_customer"].idxmax()
]

print("Best price:", best_row["price"])
print("Estimated revenue/customer:",
      best_row["revenue_per_customer"])
print("Total customers experimented on:", total_customers)
print("Total experimental revenue:", total_revenue)
results_df
```

# Experiment results plot:

```python
plt.figure(figsize=(10, 5))

plt.plot(
    results_df["price"],
    results_df["revenue_per_customer"],
    marker="o"
)

plt.axvline(
    best_row["price"],
    color="red",
    linestyle="--",
    label=f'Best observed price: {best_row["price"]:.2f}'
)

plt.xlabel("Price ()")
plt.ylabel("Estimated Revenue per Customer ()")
plt.title("Grid Search: Revenue vs. Price")
plt.legend()
plt.show()
```

# Confidence Interval code

```python
# Stage 3: Calculate confidence intervals

confidence_results = []

for price in prices:

    purchases = sales(
        price=price,
        mean_customers=1000,
        rng=rng
    )

    # Revenue generated by each customer
    customer_revenue = purchases * price

    # Average revenue per customer
    mean_revenue = customer_revenue.mean()

    # Standard error
    se = customer_revenue.std(ddof=1) / np.sqrt(len(customer_revenue))

    # 90% confidence interval
    ci_lower = mean_revenue - 1.645 * se
    ci_upper = mean_revenue + 1.645 * se

    confidence_results.append({
        "price": price,
        "mean_revenue": mean_revenue,
        "se": se,
        "ci_lower": ci_lower,
        "ci_upper": ci_upper,
        "customers": len(customer_revenue)
    })

confidence_df = pd.DataFrame(confidence_results)
confidence_df
```

## Confidence intervals for mean revenue of our top two best prices:

```python
# Find the two prices with highest estimated revenue
top_two = confidence_df.nlargest(2, "mean_revenue")
top_two
```

## Confidence interval for the difference in revenue between our top two prices:

```python
price_A = top_two.iloc[0]
price_B = top_two.iloc[1]

# Difference in estimated mean revenue
difference = (
    price_A["mean_revenue"] -
    price_B["mean_revenue"]
)

# Standard error of the difference
se_difference = np.sqrt(
    price_A["se"]**2 +
    price_B["se"]**2
)

# 90% confidence interval
ci_lower = difference - 1.645 * se_difference
ci_upper = difference + 1.645 * se_difference

print("Price A:", price_A["price"])
print("Price B:", price_B["price"])

print("Difference in revenue:", difference)
print("90% CI:", (ci_lower, ci_upper))
```

## Final independent comparison and confidence interval:

Our two best prices are 27 and 28. Now we are doing one last set of two experiments, one for each of the top two best prices. This time our experiment size is 5000.

```python
# Final independent comparison

p_A = float(price_A["price"])
p_B = float(price_B["price"])

final_size = 5000

revenue_A = sales(
    p_A,
    mean_customers=final_size,
    rng=rng
) * p_A

revenue_B = sales(
    p_B,
    mean_customers=final_size,
    rng=rng
) * p_B

# Update experiment totals
total_customers = len(revenue_A) + len(revenue_B)
total_revenue = revenue_A.sum() + revenue_B.sum()

# Difference in mean revenue
delta = revenue_A.mean() - revenue_B.mean()

# Standard error
se = np.sqrt(
    revenue_A.var(ddof=1) / len(revenue_A) +
    revenue_B.var(ddof=1) / len(revenue_B)
)

# Confidence interval
lower = delta - 1.645 * se
upper = delta + 1.645 * se

print(f"Difference: {delta:.3f}")
print(f"90% CI: ({lower:.3f}, {upper:.3f})")

if lower > 0:
    best_price = p_A
elif upper < 0:
    best_price = p_B
else:
    best_price = (
        p_A if revenue_A.mean() >= revenue_B.mean()
        else p_B
    )
    print("No statistically clear winner; selecting higher observed mean.")

print(f"Selected price: {best_price:.2f}")
```

# Final Write Up

A seller must choose one price between 0 and 70 without knowing how likely customers are to buy at each price. Every price experiment returns only a random number of customers who either buy or don’t, so any estimate of revenue per customer is noisy. Each customer in an experiment is also a real customer who pays a possibly poor price, so experimenting has a cost. The goal is to find the revenue-maximizing price while giving up as little revenue as possible along the way, measured against the benchmark of 11.925 expected profit per customer at the optimum. The difficulty is that demand has several local peaks, so a price that looks good in one region may not be the global best.       

The design narrows the price range in stages and uses larger samples as the candidates get closer together. Stage 1 is a coarse grid from 5 to 65 in 5 increments, with about 300 customers per price. It's cheap and shows which regions earn the most per customer.     

          

The best region was 20–30, so Stage 2 tested every dollar from 20 to 29 with about 500 customers each. Larger samples are used here because neighboring prices differ by small amounts that are hard to detect through noise.     



Since each customer either pays the full price or nothing, revenue per customer is highly variable without a large sample size. To quantify the uncertainty, Stage 3 ran about 1,000 customers at each price from 20 to 29 and computed 90% confidence intervals (mean ± 1.645 standard errors) for revenue per customer. We chose 90% instead of 95% or 99% because tighter confidence needs more customers, and each costs revenue.

The two best prices, 28 and 27, could not be separated, as the 90% interval for their difference (−0.14 to 1.87) included zero. Stage 4 ran a fresh, independent 5,000-customer experiment at each finalist and built an interval for the difference. The rule was to pick a price only if the interval for the difference excluded zero, and otherwise to pick the one with the higher observed mean. The result was a difference of −0.107 in the means with a 90% interval of (−0.553, 0.340), so no clear winner emerged and we selected 27.         

The first challenge was noise. Purchase counts did not always fall as price rose, and revenue per customer at neighboring prices differed by amounts comparable to random variation, so reading the best price off a single experiment was unreliable. The coarse grid that skipped by 5 was a related concern because it can skip narrow peaks, so I used it only to choose a region to examine and confirmed the finalists with larger samples. In Stage 3, 28 appeared to earn 12.65 per customer, which is above the 11.925 maximum possible expected profit. Rather than trusting that value, I re-tested the finalists on new, independent data in Stage 4, which showed that 27 had the higher mean in that experiment.        

The second challenge was deciding how much data to gether. Small samples were cheap but uncertain, while large samples were more certain but costly. We decided to use small samples to explore the data to get a better idea of what ranges would be optimal, and from there increased the sample size to determine the top 2 candidates. Once we found those we increased sample size again to determine if one price was better than another. Another challenge was interpreting the percentage of optimal figures. The benchmark of 11.925 is an expected value, so any single stage can land above or below it by chance. The final stage’s figure of 101.37% exceeds the benchmark for that reason: it covered only 27 and 28, whose true expected revenue is close to optimal, and it happened to come out about 1.5 standard errors high. I treated this as noise rather than as evidence of beating the optimum.

We experimented on 28,888 customers in total and earned 308,882 during the experiments. The optimal benchmark for that many customers is 344,489, so we captured 89.66% of the optimal profit throughout the experiment.
