1. A random sample has
   $$
   \bar{X}=52,\qquad s=10,\qquad n=100.
   $$
   Compute the estimated standard error of $\bar{X}$. Test
   $$
   H_0:\mu=50
   \qquad \text{vs.} \qquad
   H_1:\mu\ne 50
   $$
   at significance levels $\alpha=.10,.05,.01$ using the large-sample $z$-statistic. These correspond to 90%, 95%, and 99% confidence levels. Construct a 95% confidence interval for $\mu$.

$$SE_{\bar{X}} = \frac{s}{\sqrt{n}} = \frac{10}{\sqrt{100}}=1$$


$$Z = \frac{\bar{X}-\mu_0}{SE_{\bar{X}}} = \frac{52-50}{1} = 2$$

- since this is two-tailed test, we compare |z|=2 against critical values: 
    - at $\alpha = 0.1$ the critical values are 1.645 or -1.645 —— reject $H_0$
    - at $\alpha = 0.05$ the critical values are plus or minus 1.96 —— reject $H_0$
    - at $\alpha = 0.01$ the critical values are 2.576 or -2.576 —— fail to reject $H_0$ because ${2 < 2.576}$


- At 95% CI:
    $$CI = \bar{X} \pm {z_.025} \times SE_{\bar{X}} = 52 \pm 1.96 \times 1$$
    $$[50.04, 53.96]$$

2. Suppose
   $$
   \hat{S}=3.1,\qquad \widehat{SE}(\hat{S})=1.3.
   $$
   Construct 90%, 95%, and 99% confidence intervals, and test the null hypothesis that $S=0$ at the corresponding significance levels $\alpha=.10,.05,.01$. When do you reject or fail to reject the null of $S=0$?

   $$z=\frac{\hat{S}-0}{\widehat{SE}(\hat{S})}=\frac{3.1}{1.3}\approx 2.385, \qquad p\approx 0.017$$

- The confidence interval:


| Level | $z^*$ | Margin of error | Interval |
|---|---|---|---|
| 90% | 1.645 | 2.138 | $(0.962,\ 5.238)$ |
| 95% | 1.960 | 2.548 | $(0.552,\ 5.648)$ |
| 99% | 2.576 | 3.349 | $(-0.249,\ 6.449)$ |

- decision:

| $\alpha$ | Critical value | Compare $\lvert z\rvert=2.385$ | Is 0 in the CI? | Decision |
|---|---|---|---|---|
| 0.10 | 1.645 | $2.385>1.645$ | No | **Reject** $H_0$ |
| 0.05 | 1.960 | $2.385>1.960$ | No | **Reject** $H_0$ |
| 0.01 | 2.576 | $2.385<2.576$ | Yes | **Fail to reject** $H_0$ |

- since $p=0.017$, we reject $H_0: S=0$ at the 10% and 5% levels but fail to reject at the 1%. we reject exacly when the confidence interval at the maching level excludes 0.

7. For `./data/ames_prices.csv`, define an indicator equal to 1 when `Central.Air == 'Y'`. Estimate the proportion of homes with central air and test $H_0:p=.90$. Treat a proportion as the mean of a binary variable and complete the full inference routine.

```python
import numpy as np
import pandas as pd
from scipy.stats import pearsonr, bootstrap
from scipy.stats import norm
df = pd.read_csv('./data/ames_prices.csv')
df.head()
```

```python


rgn = np.random.default_rng(seed=42)
R=2000

x = (df["Central.Air"] == "Y").astype(int).to_numpy()

n=len(x)
bootstrap_estimates = np.empty(R)
for i in range(R):
    bootstrap_sample = rgn.choice(x, size=n, replace=True)
    bootstrap_estimates[i] = np.mean(bootstrap_sample)


bootstrap_se = np.std(bootstrap_estimates, ddof=1)

z_values = {
    '90%' : norm.ppf(0.95),
    '95%' : norm.ppf(0.975),
    '99%' : norm.ppf(0.995)
}

z_values
print(f"Bootstrap Standard Error: {bootstrap_se:.4f}")
```

9. For `./data/metabric.csv`, use `Overall Survival (Months)` to estimate the mean recorded time to death or last follow-up, including both living and deceased patients. Let $\mu$ denote the population mean of this recorded-time variable, and test $H_0:\mu=120$ months. Complete the full inference routine and interpret the CIs in both months and years. Living patients' eventual death times are right-censored, so this exercise does not estimate unrestricted mean lifetime. Explain that distinction in your interpretation.

```python
df3 = pd.read_csv('./data/metabric.csv')
df3.head()
```

```python

from scipy import stats

survival_times = df3['Overall Survival (Months)'].dropna().to_numpy()

n= len(survival_times)
rng = np.random.default_rng(seed=42)

mean_times = np.mean(survival_times)
std_times = np.std(survival_times, ddof=1)
se_times = std_times / np.sqrt(n)

print(f'Sample Size: {n}')
print(f"Mean Survival Time: {mean_times:.2f} months")
print(f"Standard Error: {se_times:.2f} months")
print('Standard deviation of Survival Times: {:.2f} months'.format(std_times))

t_critical = stats.t.ppf(0.975, df=n-1)
t_lower = mean_times - t_critical * se_times
t_upper = mean_times + t_critical * se_times
print(f"95% Confidence Interval for Mean Survival Time: ({t_lower:.2f}, {t_upper:.2f}) months")
```
